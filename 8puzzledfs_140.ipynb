```python
class PuzzleState:
    def __init__(self, board, parent=None, move=None, depth=0):
        self.board = board  # Tuple representation of the board for immutability
        self.parent = parent  # Parent state to reconstruct the path
        self.move = move  # Move that led to this state
        self.depth = depth  # Depth of this state in the search tree

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(self.board)

    def __str__(self):
        return '\n'.join([' '.join(map(str, row)) for row in self.board])

    def find_blank(self):
        """Finds the row and column of the blank (0) tile."""
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c

    def get_neighbors(self):
        """Generates all possible next states from the current state."""
        r_blank, c_blank = self.find_blank()
        neighbors = []
        moves = [
            (-1, 0, 'Up'),
            (1, 0, 'Down'),
            (0, -1, 'Left'),
            (0, 1, 'Right')
        ]

        for dr, dc, move_name in moves:
            new_r, new_c = r_blank + dr, c_blank + dc

            if 0 <= new_r < 3 and 0 <= new_c < 3:
                # Create a new board by swapping the blank with the new tile
                new_board_list = [list(row) for row in self.board]
                new_board_list[r_blank][c_blank], new_board_list[new_r][new_c] = \
                    new_board_list[new_r][new_c], new_board_list[r_blank][c_blank]
                new_board_tuple = tuple(tuple(row) for row in new_board_list)

                neighbors.append(PuzzleState(new_board_tuple, self, move_name, self.depth + 1))
        return neighbors
def dfs(initial_state, goal_state, max_depth=20):

    stack = [initial_state]
    visited = set()
    visited.add(initial_state.board)

    while stack:
        current_state = stack.pop()

        if current_state.board == goal_state.board:
            # Reconstruct path
            path = []
            temp = current_state
            while temp:
                path.append(temp)
                temp = temp.parent
            return path[::-1] # Reverse to get path from start to goal

        if current_state.depth < max_depth:
            for neighbor in reversed(current_state.get_neighbors()): # Reverse for DFS behavior
                if neighbor.board not in visited:
                    visited.add(neighbor.board)
                    stack.append(neighbor)

    return None


def parse_board_input(prompt):
    while True:
        print(prompt)
        board_list = []
        for i in range(3):
            row_input = input(f"Enter row {i+1} (e.g., 1 2 3): ").strip().split()
            if len(row_input) != 3:
                print("Each row must have exactly 3 numbers. Please try again.")
                break
            try:
                board_list.append(tuple(int(x) for x in row_input))
            except ValueError:
                print("Invalid input. Please enter numbers separated by spaces. Try again.")
                break
        else:
            if len(board_list) == 3:
                # Check for unique numbers 0-8
                all_numbers = [num for row in board_list for num in row]
                if sorted(all_numbers) == list(range(9)):
                    return tuple(board_list)
                else:
                    print("Invalid numbers. Board must contain unique numbers from 0 to 8. Try again.")
            else:
                print("You must enter 3 rows. Please try again.")
        print("\n") #

print("1BM24CS140")
user_initial_board = parse_board_input("Enter the initial 8-puzzle board (3 rows of 3 numbers, 0 for blank):")
user_goal_board = parse_board_input("Enter the goal 8-puzzle board (3 rows of 3 numbers, 0 for blank):")

initial_puzzle = PuzzleState(user_initial_board)
goal_puzzle = PuzzleState(user_goal_board)

print("\nInitial State Provided:")
print(initial_puzzle)
print("\nGoal State Provided:")
print(goal_puzzle)

solution_path = dfs(initial_puzzle, goal_puzzle, max_depth=30)

if solution_path:
    print(f"\nSolution Found in {len(solution_path) - 1} moves (depth: {solution_path[-1].depth}):")
    for i, state in enumerate(solution_path):
        if state.move:
            print(f"\nMove {i}: {state.move}")
        else:
            print(f"\nInitial State (Depth: {state.depth})")
        print(state)
else:
    print("\nNo solution found within the specified max_depth or puzzle is unsolvable/too complex.")
```
