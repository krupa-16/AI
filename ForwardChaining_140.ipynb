```python
class DynamicForwardChainingEngine:
    def __init__(self, patient_name, user_symptoms):
        # Format user symptoms to attach to the specific patient name
        self.facts = {f"{symptom}({patient_name})" for symptom in user_symptoms}
        self.patient = patient_name

        # The Knowledge Base rules parameterized for this specific patient
        self.rules = [
            (
                (f"Fever({self.patient})", f"SoreThroat({self.patient})"),
                f"SignsOfInfection({self.patient})"
            ),
            (
                (f"SignsOfInfection({self.patient})", f"MuscleAches({self.patient})"),
                f"TestForInfluenza({self.patient})"
            )
        ]

    def run(self):
        print(f"\n--- Starting Forward Chaining Inference ---")
        print(f"Initial Facts in KB: {sorted(list(self.facts))}\n")

        iteration = 1
        new_facts_added = True

        while new_facts_added:
            new_facts_added = False

            for premises, conclusion in self.rules:
                # Skip if the conclusion has already been derived
                if conclusion in self.facts:
                    continue

                # Check if ALL premises of the rule are met by current facts
                if all(p in self.facts for p in premises):
                    print(f"[Iteration {iteration}] Rule Triggered:")
                    print(f"   IF {' AND '.join(premises)} -> THEN {conclusion}")
                    self.facts.add(conclusion)
                    new_facts_added = True
                    iteration += 1
                    break # Restart loop with updated facts base

        print(f"\n--- Final Results for {self.patient} ---")
        print(f"Final Fact Base: {sorted(list(self.facts))}")

        target_goal = f"TestForInfluenza({self.patient})"
        if target_goal in self.facts:
            print(f"🎯 RESULT: ✅ Successfully Proved! Recommendation: {target_goal}")
        else:
            print("🎯 RESULT: ❌ Could not infer Influenza testing with the provided symptoms.")


if __name__ == "__main__":
    print("=== Medical Expert System (Forward Chaining) ===")

    # 1. Get patient name
    name = input("Enter patient name (e.g., John): ").strip()
    if not name:
        name = "Patient1"

    # 2. Get symptoms via interactive inputs
    print("\nPlease answer 'yes' or 'no' for the following symptoms:")
    symptoms_list = []

    if input("Does the patient have a fever? (y/n): ").strip().lower() in ['y', 'yes']:
        symptoms_list.append("Fever")

    if input("Does the patient have a sore throat? (y/n): ").strip().lower() in ['y', 'yes']:
        symptoms_list.append("SoreThroat")

    if input("Does the patient have muscle aches? (y/n): ").strip().lower() in ['y', 'yes']:
        symptoms_list.append("MuscleAches")

    # 3. Initialize and run engine
    engine = DynamicForwardChainingEngine(name, symptoms_list)
    engine.run()
```

```python

class DynamicCrimeEngine:
    def __init__(self, suspect, target_country, item_id):
        self.suspect = suspect
        self.country = target_country
        self.item = item_id

        # Build initial fact base dynamically based on user selections
        self.facts = set()

    def add_fact(self, fact_string):
        self.facts.add(fact_string)

    def run(self):
        print(f"\n--- Starting Forward Chaining Expert System ---")
        print(f"Initial KB Facts: {sorted(list(self.facts))}\n")

        iteration = 1
        made_inference = True

        while made_inference:
            made_inference = False

            # Rule 1: Missile(x) => Weapon(x)
            if f"Missile({self.item})" in self.facts and f"Weapon({self.item})" not in self.facts:
                print(f"[Iteration {iteration}] Rule Triggered: Missile({self.item}) => Weapon({self.item})")
                self.facts.add(f"Weapon({self.item})")
                made_inference = True

            # Rule 2: Enemy(x, America) => Hostile(x)
            if f"Enemy({self.country}, America)" in self.facts and f"Hostile({self.country})" not in self.facts:
                print(f"[Iteration {iteration}] Rule Triggered: Enemy({self.country}, America) => Hostile({self.country})")
                self.facts.add(f"Hostile({self.country})")
                made_inference = True

            # Rule 3: Missile(x) AND Owns(Country, x) => Sells(Suspect, x, Country)
            if f"Missile({self.item})" in self.facts and f"Owns({self.country}, {self.item})" in self.facts:
                sales_fact = f"Sells({self.suspect}, {self.item}, {self.country})"
                if sales_fact not in self.facts:
                    print(f"[Iteration {iteration}] Rule Triggered: Missile({self.item}) ∧ Owns({self.country}, {self.item}) => Sells...")
                    self.facts.add(sales_fact)
                    made_inference = True

            # Rule 4: American(p) AND Weapon(q) AND Sells(p, q, r) AND Hostile(r) => Criminal(p)
            if (f"American({self.suspect})" in self.facts and
                f"Weapon({self.item})" in self.facts and
                f"Sells({self.suspect}, {self.item}, {self.country})" in self.facts and
                f"Hostile({self.country})" in self.facts):

                crime_fact = f"Criminal({self.suspect})"
                if crime_fact not in self.facts:
                    print(f"[Iteration {iteration}] Rule Triggered: Full conditions met! => {crime_fact}")
                    self.facts.add(crime_fact)
                    made_inference = True

            if made_inference:
                iteration += 1

        print(f"\n--- Final Verdict ---")
        print(f"Final Facts in KB: {sorted(list(self.facts))}")

        goal = f"Criminal({self.suspect})"
        if goal in self.facts:
            print(f"🎯 RESULT: ✅ {self.suspect} is legally classified as a criminal under FOL rules.")
        else:
            print(f"🎯 RESULT: ❌ Innocent! Insufficient evidence to prove {self.suspect} is a criminal.")


if __name__ == "__main__":
    print("=== International Arms Trafficking Expert System ===")

    # 1. Gather custom entity names
    suspect_name = input("Enter suspect's name (e.g., Robert): ").strip() or "Robert"
    foreign_nation = input("Enter enemy nation name (e.g., A): ").strip() or "A"
    weapon_serial = input("Enter missile asset identifier (e.g., T1): ").strip() or "T1"

    engine = DynamicCrimeEngine(suspect_name, foreign_nation, weapon_serial)

    # 2. Dynamically build the premise state using input queries
    print("\n--- Constructing the Case (Answer True/False) ---")

    if input(f"Is {suspect_name} an American citizen? (y/n): ").lower() in ['y', 'yes']:
        engine.add_fact(f"American({suspect_name})")

    if input(f"Is country {foreign_nation} considered an enemy of America? (y/n): ").lower() in ['y', 'yes']:
        engine.add_fact(f"Enemy({foreign_nation}, America)")

    if input(f"Does country {foreign_nation} possess asset {weapon_serial}? (y/n): ").lower() in ['y', 'yes']:
        engine.add_fact(f"Owns({foreign_nation}, {weapon_serial})")

    if input(f"Is asset {weapon_serial} a missile payload? (y/n): ").lower() in ['y', 'yes']:
        engine.add_fact(f"Missile({weapon_serial})")

    # 3. Process the reasoning cycle
    engine.run()
```
