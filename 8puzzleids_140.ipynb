```python
def get_neighbors(state):
    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)

    moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]

    for dr, dc, move_name in moves:
        new_row, new_col = row + dr, col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero_idx = new_row * 3 + new_col

            state_list = list(state)
            state_list[zero_idx], state_list[new_zero_idx] = state_list[new_zero_idx], state_list[zero_idx]
            neighbors.append((tuple(state_list), move_name))

    return neighbors

def depth_limited_search(start_state, goal_state, limit):
    stack = [(start_state, [], 0)]
    visited = {}
    cutoff_occurred = False

    while stack:
        current_state, path, depth = stack.pop()

        if current_state == goal_state:
            return path

        if depth >= limit:
            if depth == limit:
                cutoff_occurred = True
            continue

        if current_state in visited and visited[current_state] <= depth:
            continue

        visited[current_state] = depth

        for neighbor, move_name in get_neighbors(current_state):
            stack.append((neighbor, path + [move_name], depth + 1))

    return 'cutoff' if cutoff_occurred else None

def iterative_deepening_search(start_state, goal_state):
    depth = 0
    while True:
        result = depth_limited_search(start_state, goal_state, depth)
        if result != 'cutoff':
            return result
        depth += 1

def print_board(state):
    for i in range(0, 9, 3):
        print(list(state[i:i+3]))
    print()

def get_user_input(prompt_text):
    while True:
        try:
            print(prompt_text)
            user_in = input("Enter 9 numbers (0-8) separated by spaces: ")
            parsed = [int(x) for x in user_in.split()]

            if len(parsed) != 9 or set(parsed) != set(range(9)):
                print("Invalid input! You must enter exactly 9 unique digits from 0 to 8.\n")
                continue

            return tuple(parsed)
        except ValueError:
            print("Invalid characters encountered. Please enter numbers only.\n")

if __name__ == "__main__":
    start = get_user_input("Enter the START board state:")
    goal = get_user_input("\nEnter the GOAL board state:")

    print("\nInitial Configuration:")
    print_board(start)
    print("Target Configuration:")
    print_board(goal)

    solution = iterative_deepening_search(start, goal)

    if solution is not None:
        print(f"Solved in {len(solution)} moves.")
        print(f"Move Path: {solution}")
    else:
        print("Could not find a path to the goal.")
```
