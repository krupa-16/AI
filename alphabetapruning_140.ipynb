```python
import math

class Node:
    def __init__(self, name, value=None):
        self.name = name
        self.value = value
        self.children = []

def build_tree(name="Root"):
    is_leaf = input(f"Is '{name}' a leaf node? (y/n): ").strip().lower()

    if is_leaf == 'y':
        val = int(input(f"  Enter integer value for leaf '{name}': "))
        return Node(name, value=val)

    node = Node(name)
    num_children = int(input(f"  How many children does '{name}' have? (e.g., 1, 2, 3): "))

    for i in range(1, num_children + 1):
        child_name = f"{name}.{i}"
        node.children.append(build_tree(child_name))

    return node

def alpha_beta_pruning(node, is_maximizing, alpha, beta):
    # Base case: reached a leaf node (no children)
    if not node.children:
        return node.value

    if is_maximizing:
        best = -math.inf
        for child in node.children:
            val = alpha_beta_pruning(child, False, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                print(f"  [Pruned] Remaining children of '{node.name}' skipped (alpha={alpha}, beta={beta})")
                break
        return best
    else:
        best = math.inf
        for child in node.children:
            val = alpha_beta_pruning(child, True, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                print(f"  [Pruned] Remaining children of '{node.name}' skipped (alpha={alpha}, beta={beta})")
                break
        return best

# --- Main Execution ---
if __name__ == "__main__":
    try:
        print("--- Build Your Game Tree ---")
        root = build_tree()

        choice = input("\nDoes the Root player maximize? (y/n): ").strip().lower()
        is_maximizing = (choice == 'y')

        print("\n--- Running Alpha-Beta Pruning ---")
        optimal_value = alpha_beta_pruning(root, is_maximizing, -math.inf, math.inf)
        print(f"\nThe optimal value is: {optimal_value}")

    except ValueError:
        print("Invalid input. Please enter valid integers for node values and child counts.")
```
