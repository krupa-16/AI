```python
def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbors(state):
    neighbors = []

    for i in range(len(state)):
        for j in range(i + 1, len(state)):
            new_state = state.copy()
            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append((new_state, calculate_cost(new_state), i, j))

    return neighbors


def hill_climbing(initial_state):
    current = initial_state.copy()

    while True:
        current_cost = calculate_cost(current)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("\nGoal State Reached!")
            return current

        neighbors = generate_neighbors(current)

        print("\nNeighbours:")

        for state, cost, i, j in neighbors:
            print(
                f"Swap (x{i}, x{j}) -> {state}  Cost = {cost}"
            )

        best_cost = min(cost for state, cost, i, j in neighbors)

        best_neighbors = [
            item for item in neighbors if item[1] == best_cost
        ]

        best_neighbor = best_neighbors[0]

        for item in best_neighbors:
            if (item[2], item[3]) < (best_neighbor[2], best_neighbor[3]):
                best_neighbor = item

        next_state, next_cost, i, j = best_neighbor

        if next_cost >= current_cost:
            print("\nNo better neighbour found.")
            return current

        print(
            f"\nSelected: Swap (x{i}, x{j}) -> "
            f"{next_state}, Cost = {next_cost}"
        )

        current = next_state


initial_state = [3, 1, 2, 0]

solution = hill_climbing(initial_state)

print("\nFinal Solution:", solution)
print("Final Cost:", calculate_cost(solution))
```
