```python
import itertools

def get_alpha(A, B):
    return A or B

def get_kb(A, B, C):
    return (A or C) and (B or (not C))

combinations = list(itertools.product([False, True], repeat=3))

print("A B C | A∨C B∨¬C | KB α | Model Status")
print("-" * 42)

entails = True

for A, B, C in combinations:
    a_or_c = A or C
    b_or_not_c = B or (not C)
    kb = get_kb(A, B, C)
    alpha = get_alpha(A, B)

    if kb and not alpha:
        entails = False

    status = "<-- KB is TRUE" if kb else ""
    print(f"{int(A)} {int(B)} {int(C)} |  {int(a_or_c)}    {int(b_or_not_c)}  |  {int(kb)}  {int(alpha)} | {status}")

print("\n" + "="*30)
print(f"Conclusion: KB ⊨ α is {entails}")
print("="*30)
```
