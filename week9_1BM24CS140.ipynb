```python

def fol_resolution():
    print("FIRST ORDER LOGIC - RESOLUTION")
    print("--------------------------------")

    print("\nEnter the given facts (yes/no):")

    likes_food = input("1. Does John like all food? ").lower()
    apple = input("2. Is Apple food? ").lower()
    vegetables = input("3. Are Vegetables food? ").lower()
    eats = input("4. Does Anil eat peanuts? ").lower()
    alive = input("5. Is Anil alive? ").lower()

    print("\n--- RESOLUTION PROCESS ---")

    if likes_food == "yes" and eats == "yes" and alive == "yes":

        print("\nStep 1: Alive(Anil)")
        print("Step 2: Alive(x) -> Not Killed(x)")
        print("Step 3: Therefore, Not Killed(Anil)")

        print("\nStep 4: Eats(Anil, Peanuts)")
        print("Step 5: Eats(x,y) AND Not Killed(x) -> Food(y)")
        print("Step 6: Therefore, Food(Peanuts)")

        print("\nStep 7: John likes all food")
        print("Step 8: Food(Peanuts) -> Likes(John, Peanuts)")

        print("\n--- RESOLUTION GRAPH ---\n")

        print("          Alive(Anil)")
        print("                |")
        print("                v")
        print("        Not Killed(Anil)")
        print("                |")
        print("    Eats(Anil, Peanuts)")
        print("                |")
        print("                v")
        print("         Food(Peanuts)")
        print("                |")
        print("       John likes all food")
        print("                |")
        print("                v")
        print("    Likes(John, Peanuts)")
        print("                |")
        print("          Negated Goal")
        print("                |")
        print("                v")
        print("               NIL")

        print("\nCONCLUSION:")
        print("John likes peanuts is PROVED using FOL Resolution.")

    else:
        print("\nCONCLUSION:")
        print("Required facts are missing.")
        print("Goal cannot be proved.")

fol_resolution()
```

```python

def resolution():
    print("FOL RESOLUTION")
    print("-------------------------")

    science = input("Does Alice like Science? (yes/no): ").lower()
    bob = input("Does Bob like Physics? (yes/no): ").lower()
    rule1 = input("Does liking Science imply Physics or Mathematics? (yes/no): ").lower()
    rule2 = input("Does liking Mathematics imply Statistics? (yes/no): ").lower()
    rule3 = input("Does liking Statistics imply Data Analysis? (yes/no): ").lower()

    if (science == "yes" and rule1 == "yes"
        and rule2 == "yes" and rule3 == "yes"):

        print("\nRESOLUTION STEPS")

        print("1. Alice likes Science.")
        print("2. Alice likes Physics OR Mathematics.")
        print("3. Alice likes Mathematics.")
        print("4. Alice likes Statistics.")
        print("5. Alice likes Data Analysis.")
        print("6. Resolve with negated goal.")
        print("7. NIL - Contradiction found.")

        print("\nRESOLUTION GRAPH")
        print("Likes(Alice, Science)")
        print("          |")
        print("          v")
        print("Physics OR Mathematics")
        print("          |")
        print("          v")
        print("Likes(Alice, Mathematics)")
        print("          |")
        print("          v")
        print("Likes(Alice, Statistics)")
        print("          |")
        print("          v")
        print("Likes(Alice, Data Analysis)")
        print("          |")
        print("          v")
        print("         NIL")

        print("\nCONCLUSION:")
        print("Alice likes Data Analysis is PROVED.")

    else:
        print("\nCannot prove the given statement.")

resolution()
```
