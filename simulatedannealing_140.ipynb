```python

import random
import math

def cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts

def display(state):
    n = len(state)
    for row in range(n):
        for col in range(n):
            print("Q" if state[col] == row else ".", end=" ")
        print()

def simulated_annealing(current):
    n = len(current)
    temperature = 1000.0
    cooling_rate = 0.995

    while temperature > 0.01:
        if cost(current) == 0:
            return current

        next_state = current.copy()
        col = random.randrange(n)
        next_state[col] = random.choice(
            [r for r in range(n) if r != current[col]]
        )

        delta = cost(current) - cost(next_state)

        if delta > 0 or random.random() < math.exp(delta / temperature):
            current = next_state

        temperature *= cooling_rate

    return current

n = int(input("Enter number of queens: "))
state = list(map(int, input("Enter queen row positions: ").split()))

if n < 2 or len(state) != n or any(x < 1 or x > n for x in state):
    print("Invalid input")
else:
    state = [x - 1 for x in state]
    solution = simulated_annealing(state)

    print("Final board:")
    display(solution)
    print("Conflicts:", cost(solution))

    if cost(solution) == 0:
        print("Solution found")
    else:
        print("Solution not found. Run again.")
```
